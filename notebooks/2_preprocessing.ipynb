```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv(r'C:\Users\shreyaash mogaveera\MedBot\data\dataset.csv')
```

```python
#removing space from symptom names
symptom_cols=[col for col in df.columns if col.startswith('Symptom_')]

for col in symptom_cols:
    df[col]=df[col].str.strip()
    df[col]=df[col].str.replace(' ','_')
    df[col]=df[col].str.lower()
```

```python
#getting all the unique symptoms
all_symptoms=set()
for col in symptom_cols:
    all_symptoms.update(df[col].dropna().unique())
all_symptoms=sorted(list(all_symptoms))

print("total unique symptoms:",len(all_symptoms))
print("\n Sample symptoms:",all_symptoms[:10])
```

```python
#creating binary feature matrix
#1=present, 0=absent
X=pd.DataFrame(0,index=df.index,columns=all_symptoms)
for col in symptom_cols:
    for idx,symptom in df[col].items():
        if pd.notna(symptom) and symptom in X.columns:
            X.loc[idx,symptom]=1

print("feature matrix shape:",X.shape)
print("\nSample") 
X.sample(5)
```

```python
#encode the target variable
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
y=le.fit_transform(df['Disease'])

print("number of classes:", len(le.classes_))
print("classes:",le.classes_)
```

```python
import joblib
joblib.dump(all_symptoms,'../model/symptom_columns.pkl')
joblib.dump(le,'../model/label_encoder.pkl')
```
